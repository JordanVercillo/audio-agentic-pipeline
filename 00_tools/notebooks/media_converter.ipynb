# Universal Media Converter

This notebook uses `media_converter.py` (same folder) to convert media files.

| Output type | Formats | Notes |
|-------------|---------|-------|
| **Audio** | `mp3`, `wav` | Extracts audio from video, or converts between audio formats |
| **Video** | `mp4`, `avi`, `mov`, `mkv` | Optional `video_resolution` -- omit to keep original, or e.g. `(1920, 1080)` |

```python
from media_converter import convert_media
```

---
### Example A -- Local folder to audio (MP3 or WAV)

```python
convert_media(
    input_source  = r"C:\Users\jverc\Downloads",
    input_format  = "mp4",
    output_format = "wav",
    prefix        = "wav_",
)
```

### Example B -- YouTube URL to audio

```python
convert_media(
    input_source  = "https://www.youtube.com/watch?v=t5bsh9NRHMY",
    output_format = "wav",
    output_folder = r"C:\Users\jverc\Downloads",
    prefix        = "",
)
```

### Example C -- Local folder to video (with resolution)

```python
#Resoultions - 4K: (2560, 1440), 1080p: (1920,1080), 720p: (1280, 720)

convert_media(
    input_source     = r"C:\Users\jverc\Downloads",
    input_format     = "mp4",
    output_format    = "mp4",
    video_resolution = (1920, 1080),
    prefix           = "hd_",
)
```

### Example D -- YouTube URL to video (with resolution)

```python
#Resoultions - 4K: (3840, 2160) 1440p: (2560, 1440), 1080p: (1920,1080), 720p: (1280, 720)

convert_media(
    input_source     = "https://www.youtube.com/watch?v=EkHTsc9PU2A",
    output_format    = "mp4",
    output_folder    = r"C:\Users\jverc\Downloads",
    video_resolution = (2560, 1440),
    prefix           = "yt_1440p_",
)
```

### Example E -- YouTube URL to video (keep original resolution)

```python
# No video_resolution = keeps the original quality (e.g. 4K stays 4K)

convert_media(
    input_source  = "https://www.youtube.com/watch?v=EkHTsc9PU2A",
    output_format = "mp4",
    output_folder = r"C:\Users\jverc\Downloads",
    prefix        = "yt_original_",
)
```
